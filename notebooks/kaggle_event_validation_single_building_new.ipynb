# Kaggle Energy Anomaly Detection — Single Building

This notebook runs the anomaly detector on **one `building_id`** and evaluates/visualizes predictions on that building only.



```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

from anomlib.datasets.kaggle_energy.load import load_train
from anomlib.detectors import EnergyTimeSeriesDetector, GenericTimeSeriesDetector

try:
    from anomlib.detectors import EnergySupervisedDetector
except Exception:
    EnergySupervisedDetector = None

from anomlib.core.eventing import scores_to_events
```

```python

# Pick ONE building to analyze
BUILDING_ID = 1068

# Train/test split within this building's timeline
TRAIN_FRAC = 0.8

CANDIDATE_DATA_PATHS = [
    "data/kaggle/train.csv",
    "../data/kaggle/train.csv",
    "../../data/kaggle/train.csv",
]

# Choose detector mode
DETECTOR_TYPE = "unsupervised"   # "unsupervised" or "supervised"
UNSUPERVISED_DETECTOR = "energy" # "energy" or "generic"

# Unsupervised knobs 
DIRECTION = "both"
THRESHOLD = 6.0
THRESHOLD_QUANTILE = 0.9995
THRESHOLD_CAP = 10.0
THRESHOLD_END_RATIO = 0.25
MIN_DURATION = "2h"
GAP_TOLERANCE = "15min"

# Supervised knobs (only used if DETECTOR_TYPE == "supervised")
SUPERVISED_THRESHOLD = 0.9
SUPERVISED_THRESHOLD_STRATEGY = "recall_target"  
SUPERVISED_RECALL_TARGET = 0.65
```

```python
# Load dataset 
def resolve_data_path(candidates):
    # Try relative to current working directory first
    for p in candidates:
        if Path(p).exists():
            return str(Path(p))
    # Try relative to the notebook file location if running from repo root is not true
    for p in candidates:
        ap = (Path.cwd() / p).resolve()
        if ap.exists():
            return str(ap)
    raise FileNotFoundError(
        "Could not find train.csv. Tried: " + ", ".join(candidates) +
        "\nPut the file at data/kaggle/train.csv (repo root) or update CANDIDATE_DATA_PATHS."
    )

DATA_PATH = resolve_data_path(CANDIDATE_DATA_PATHS)
print("Using DATA_PATH:", DATA_PATH)

df = load_train(DATA_PATH)
df["timestamp"] = pd.to_datetime(df["timestamp"])
df = df.dropna(subset=["meter_reading"]).reset_index(drop=True)

# Filter to one building
b = (
    df[df["building_id"] == BUILDING_ID]
    .copy()
    .sort_values("timestamp")
    .reset_index(drop=True)
)

print("rows:", len(b), "time span:", b["timestamp"].min(), "->", b["timestamp"].max())
b.head()
```

```python
# Per-building time split 
def time_split(df_in: pd.DataFrame, frac: float):
    df_in = df_in.sort_values("timestamp").reset_index(drop=True)
    k = int(len(df_in) * frac)
    return df_in.iloc[:k].copy(), df_in.iloc[k:].copy()

train_b, test_b = time_split(b, TRAIN_FRAC)
print("train rows:", len(train_b), "test rows:", len(test_b))
```

```python
# Fit + detect 
if DETECTOR_TYPE == "supervised":
    if EnergySupervisedDetector is None:
        raise ImportError("EnergySupervisedDetector could not be imported from anomlib.detectors in this environment.")
    if "anomaly" not in train_b.columns:
        raise ValueError("Supervised detector requires an 'anomaly' column in the dataset.")
    det = EnergySupervisedDetector(
        entity_col="building_id",
        time_col="timestamp",
        value_col="meter_reading",
        threshold=SUPERVISED_THRESHOLD,
        threshold_strategy=SUPERVISED_THRESHOLD_STRATEGY,
        recall_target=SUPERVISED_RECALL_TARGET,
    )
    fit_b = train_b.copy()
    print("Fitting supervised on all train rows:", len(fit_b))
else:
    # Unsupervised: fit on normal-only rows if labels exist
    if "anomaly" in train_b.columns:
        fit_b = train_b[train_b["anomaly"] == 0].copy()
        print("Fitting unsupervised on normal-only rows:", len(fit_b), "/", len(train_b))
    else:
        fit_b = train_b.copy()
        print("Fitting unsupervised on all train rows:", len(fit_b))

    Detector = EnergyTimeSeriesDetector if UNSUPERVISED_DETECTOR == "energy" else GenericTimeSeriesDetector
    det = Detector(
        entity_col="building_id",
        time_col="timestamp",
        value_col="meter_reading",
        direction=DIRECTION,
        threshold=THRESHOLD,
        threshold_quantile=THRESHOLD_QUANTILE,
        threshold_cap=THRESHOLD_CAP,
        threshold_end_ratio=THRESHOLD_END_RATIO,
        min_duration=MIN_DURATION,
        gap_tolerance=GAP_TOLERANCE,
    )

det.fit(fit_b)
events, scores = det.detect(test_b)
print("events:", len(events))
events[:3]
```

```python
# Attach score to test rows 
df_eval = test_b.copy().reset_index(drop=True)

if isinstance(scores, pd.DataFrame) and "score" in scores.columns:
    df_eval["score"] = scores["score"].to_numpy()
elif isinstance(scores, pd.Series):
    df_eval["score"] = scores.to_numpy()
else:
    df_eval["score"] = np.asarray(scores)

df_eval.head()
```

```python
#Convert event intervals -> pointwise predictions 
def label_points_from_events_single(df_in: pd.DataFrame, events) -> np.ndarray:
    if not events:
        return np.zeros(len(df_in), dtype=np.int8)
    intervals = [(pd.to_datetime(e.start), pd.to_datetime(e.end)) for e in events]
    intervals.sort(key=lambda x: x[0])

    t = df_in["timestamp"].to_numpy()
    j = 0
    m = np.zeros(len(df_in), dtype=np.int8)

    for i in range(len(df_in)):
        ti = t[i]
        while j < len(intervals) and intervals[j][1] < ti:
            j += 1
        if j < len(intervals):
            s, e = intervals[j]
            if s <= ti <= e:
                m[i] = 1
    return m

df_eval["pred"] = label_points_from_events_single(df_eval, events)
df_eval["pred"].value_counts()
```

```python
#  True events from labels (optional, for visualization) 
if "anomaly" in df_eval.columns:
    true_events = scores_to_events(
        df=df_eval,
        score_col="anomaly",
        entity_col="building_id",
        time_col="timestamp",
        direction="high",
        threshold=0.5,
        min_duration=pd.Timedelta("1h"),
        gap_tolerance=pd.Timedelta("1h"),
    )
else:
    true_events = []
```

```python
#  Pointwise metrics (if labels exist) 
if "anomaly" in df_eval.columns:
    tp = int(((df_eval["pred"] == 1) & (df_eval["anomaly"] == 1)).sum())
    fp = int(((df_eval["pred"] == 1) & (df_eval["anomaly"] == 0)).sum())
    fn = int(((df_eval["pred"] == 0) & (df_eval["anomaly"] == 1)).sum())

    precision = tp / (tp + fp) if (tp + fp) else 0.0
    recall = tp / (tp + fn) if (tp + fn) else 0.0
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) else 0.0

    print("tp fp fn:", tp, fp, fn)
    print(f"precision={precision:.3f}  recall={recall:.3f}  f1={f1:.3f}")
else:
    print("No anomaly labels in data — skipping metrics.")
```

```python
# Visualize: meter_reading + score + shaded predicted/true events 
def plot_building(df_in: pd.DataFrame, pred_events, true_events=None):
    # meter_reading
    fig = plt.figure(figsize=(14, 7))
    ax1 = plt.gca()
    ax1.plot(df_in["timestamp"], df_in["meter_reading"], linewidth=1)
    ax1.set_title(f"Building {BUILDING_ID} — meter_reading (test split)")
    ax1.set_xlabel("timestamp")
    ax1.set_ylabel("meter_reading")

    # shade predicted events
    for k, e in enumerate(pred_events):
        ax1.axvspan(pd.to_datetime(e.start), pd.to_datetime(e.end), alpha=0.15,
                    label="Predicted" if k == 0 else None)

    # shade true events
    if true_events:
        for k, e in enumerate(true_events):
            ax1.axvspan(pd.to_datetime(e.start), pd.to_datetime(e.end), alpha=0.10,
                        color="red", label="True Anomaly" if k == 0 else None)

    ax1.legend()
    plt.show()

    # score
    fig = plt.figure(figsize=(14, 5))
    ax2 = plt.gca()
    ax2.plot(df_in["timestamp"], df_in["score"], linewidth=1)
    ax2.set_title(f"Building {BUILDING_ID} — anomaly score (test split)")
    ax2.set_xlabel("timestamp")
    ax2.set_ylabel("score")

    for e in pred_events:
        ax2.axvspan(pd.to_datetime(e.start), pd.to_datetime(e.end), alpha=0.15)
    if true_events:
        for e in true_events:
            ax2.axvspan(pd.to_datetime(e.start), pd.to_datetime(e.end), alpha=0.10, color="red")

    plt.show()

plot_building(df_eval, events, true_events)
```

```python
# Inspect events table 
def events_table(events):
    rows = []
    for i, e in enumerate(events):
        rows.append({
            "event_idx": i,
            "start": pd.to_datetime(e.start),
            "end": pd.to_datetime(e.end),
            "duration_h": (pd.to_datetime(e.end) - pd.to_datetime(e.start)).total_seconds() / 3600.0,
            "direction": getattr(e, "direction", None),
            "score_peak": getattr(e, "score_peak", np.nan),
            "score_mean": getattr(e, "score_mean", np.nan),
            "severity": getattr(e, "severity", np.nan),
            "reason": getattr(e, "reason", ""),
        })
    return pd.DataFrame(rows)

evt_df = events_table(events)
evt_df.head(30)
```
